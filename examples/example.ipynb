```python
from ipyd3geo import GeoJSON, Map, Projection, VectorStyle
```

```python
m = Map(
    projection="geoAzimuthalEqualArea",
    border=True,
    graticule=True,
    dynamic=False,
    basemap=True,
    hamburger=False,
)
m
```

```python
projection = Projection(
    "geoConicEqualArea",
    rotate=(-30, 0, 0),
    center=(0.0, 0.0),
    precision=0.5,
    parallels=(20, 50),
)
m2 = Map(
    projection=projection, border=True, graticule=True, dynamic=False, basemap=True
)
m2
```

```python
m3 = Map(
    projection="geoEqualEarth", border=True, graticule=True, dynamic=False, basemap=True
)
m3.add(
    GeoJSON(
        data="amerisquare.geojson",
        name="amerisquare",
        style=VectorStyle(
            stroke_color="#ff0000",
            fill_color="#00ff00",
            weight=5,
            stroke_opacity=0.8,
            fill_opacity=0.5,
        ),
    )
)
m3
```

```python
m4 = Map(
    projection="geoOrthographic",
    graticule=True,
    basemap=True,
    dynamic="rotate",
)
m4
```

```python
m5 = Map(
    projection=Projection(
        "geoPolyhedralWaterman",
    ),
    basemap=True,
    graticule=True,
    dynamic=False,
)
m5
```
