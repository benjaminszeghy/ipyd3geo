```python
from ipyd3geo import Map
```

```python
m=Map(projection="geoAzimuthalEqualArea", border=True, graticule=True)
m
```

```python
from ipyd3geo import Projection
projection=Projection(
    "geoConicEqualArea", 
    rotate=(-30, 0, 0), 
    center=(0.0, 0.0), 
    precision=0.5, 
    parallels=(20, 50)
    )
m2 = Map(projection=projection, border=True, graticule=True)
m2
```
